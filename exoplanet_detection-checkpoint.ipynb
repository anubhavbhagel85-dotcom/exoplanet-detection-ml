# 🌌 Exoplanet Detection from Stellar Light Curves
### Using NASA Kepler Mission Data + Machine Learning

---

## What are we doing here?

When a planet orbits a star and passes **in front of it** (called a **transit**), it blocks a tiny fraction of the star's light. If we measure a star's brightness over time, we can see periodic dips — that's our signal.

NASA's **Kepler Space Telescope** measured brightness (flux) of ~150,000 stars continuously for 4 years. This notebook uses that data to build an ML model that classifies stars as:
- **Class 2** → Confirmed exoplanet (planet found!)
- **Class 1** → Not an exoplanet

## Step 4: Preprocessing

### Why preprocessing matters:
1. **Normalization** — stars have different base brightness levels. We normalize so the model focuses on the *shape* of the curve, not the absolute values.
2. **Class imbalance** — there are ~5000 normal stars but only ~37 exoplanet stars. If we don't fix this, the model will just predict "no planet" for everything and get 99% accuracy (but be useless).

```python
# ── Separate features and labels ──
X_train = train_df.drop('LABEL', axis=1).values
y_train = train_df['LABEL'].values - 1   # convert: 2→1 (exoplanet), 1→0 (not)

X_test  = test_df.drop('LABEL', axis=1).values
y_test  = test_df['LABEL'].values - 1

print(f'X_train shape: {X_train.shape}  (5087 stars, 3197 time steps each)')
print(f'X_test shape:  {X_test.shape}')
print(f'Training labels — Exoplanet: {y_train.sum()}  |  Not exoplanet: {(y_train==0).sum()}')
```

```python
# ── Normalize each light curve individually (row-wise) ──
# We subtract the mean and divide by std — this is called "standardization"
# It makes every star's curve comparable regardless of base brightness

def normalize_lightcurves(X):
    mean = X.mean(axis=1, keepdims=True)
    std  = X.std(axis=1, keepdims=True)
    std[std == 0] = 1   # avoid division by zero
    return (X - mean) / std

X_train_norm = normalize_lightcurves(X_train)
X_test_norm  = normalize_lightcurves(X_test)

print('Normalization done!')
print(f'Before: mean={X_train[0].mean():.2f}, std={X_train[0].std():.2f}')
print(f'After:  mean={X_train_norm[0].mean():.4f}, std={X_train_norm[0].std():.4f}')
```

```python
# ── Handle class imbalance using OVERSAMPLING ──
# We duplicate exoplanet examples (minority class) to balance the dataset
# This is called "upsampling" the minority class

# Combine for resampling
X_combined = np.vstack([X_train_norm])
y_combined = y_train.copy()

# Separate classes
X_majority = X_combined[y_combined == 0]
X_minority = X_combined[y_combined == 1]
y_majority = y_combined[y_combined == 0]
y_minority = y_combined[y_combined == 1]

# Oversample minority class to match majority
X_minority_upsampled, y_minority_upsampled = resample(
    X_minority, y_minority,
    replace=True,
    n_samples=len(X_majority),
    random_state=42
)

# Combine back
X_balanced = np.vstack([X_majority, X_minority_upsampled])
y_balanced = np.hstack([y_majority, y_minority_upsampled])

# Shuffle
shuffle_idx = np.random.permutation(len(X_balanced))
X_balanced  = X_balanced[shuffle_idx]
y_balanced  = y_balanced[shuffle_idx]

print(f'Before balancing — Exoplanet: {y_train.sum()} | Not: {(y_train==0).sum()}')
print(f'After balancing  — Exoplanet: {y_balanced.sum()} | Not: {(y_balanced==0).sum()}')
print('Now the model sees equal examples of both classes!')
```

## Step 5: Baseline Model — Random Forest

Before building a complex neural network, always try a simpler model first. This gives us a **baseline** to beat.

Random Forest works on features we extract from the light curves — not the raw time series directly.

```python
# ── Extract statistical features from light curves ──
# Instead of giving the model 3197 raw time steps,
# we compute meaningful summaries of each curve

def extract_features(X):
    features = []
    for curve in X:
        feat = [
            np.mean(curve),            # average brightness
            np.std(curve),             # spread of brightness
            np.min(curve),             # minimum (deepest dip)
            np.max(curve),             # maximum
            np.max(curve) - np.min(curve),  # range (peak to trough)
            np.percentile(curve, 25),  # 25th percentile
            np.percentile(curve, 75),  # 75th percentile
            np.median(curve),          # median
            np.sum(curve < np.mean(curve) - 2*np.std(curve)),  # count of big dips
            np.mean(np.abs(np.diff(curve))),  # average rate of change
            np.var(curve),             # variance
            # Skewness (asymmetry) — transits cause negative skew
            pd.Series(curve).skew(),
            # Kurtosis (heaviness of tails)
            pd.Series(curve).kurt(),
        ]
        features.append(feat)
    return np.array(features)

print('Extracting features from training data...')
X_train_feat = extract_features(X_balanced)
X_test_feat  = extract_features(X_test_norm)
print(f'Feature matrix shape: {X_train_feat.shape}')  # should be (n_stars, 13)
print('Features extracted!')
```

```python
# ── Train Random Forest ──
print('Training Random Forest...')
rf_model = RandomForestClassifier(
    n_estimators=200,     # 200 decision trees
    max_depth=10,         # don't let trees get too deep (prevents overfitting)
    random_state=42,
    n_jobs=-1             # use all CPU cores
)
rf_model.fit(X_train_feat, y_balanced)
print('Random Forest trained!')
```

```python
# ── Evaluate Random Forest ──
rf_pred      = rf_model.predict(X_test_feat)
rf_pred_prob = rf_model.predict_proba(X_test_feat)[:, 1]

print('='*50)
print('RANDOM FOREST RESULTS')
print('='*50)
print(classification_report(y_test, rf_pred, 
                              target_names=['No Planet', 'Exoplanet']))

rf_roc = roc_auc_score(y_test, rf_pred_prob)
rf_f1  = f1_score(y_test, rf_pred)
print(f'ROC-AUC Score: {rf_roc:.4f}  (1.0 = perfect, 0.5 = random guessing)')
print(f'F1 Score:      {rf_f1:.4f}')

# Confusion matrix
fig, ax = plt.subplots(figsize=(6, 5))
cm = confusion_matrix(y_test, rf_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Planet', 'Exoplanet'],
            yticklabels=['No Planet', 'Exoplanet'])
ax.set_title('Random Forest — Confusion Matrix', fontsize=13, fontweight='bold')
ax.set_ylabel('Actual')
ax.set_xlabel('Predicted')
plt.tight_layout()
plt.savefig('rf_confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
print('\nReading the matrix:')
print('  Top-left = correctly said "no planet"')
print('  Bottom-right = correctly detected exoplanets')
print('  Bottom-left = missed exoplanets (false negatives — most costly!)')
```

## Step 6: Deep Learning Model — 1D Convolutional Neural Network (CNN)

### Why CNN works on time series:
A 1D CNN slides a "filter" across the time axis and learns to **detect local patterns** — like the characteristic dip shape of a transit. It's the same idea as 2D CNNs detecting edges in images, but on a 1D signal.

This is exactly the kind of model real astroinformatics researchers use.

```python
# ── Prepare data for CNN ──
# CNNs need shape: (n_samples, timesteps, channels)
# We have 1 channel (just flux) so we add a dimension

X_train_cnn = X_balanced.reshape(X_balanced.shape[0], X_balanced.shape[1], 1)
X_test_cnn  = X_test_norm.reshape(X_test_norm.shape[0], X_test_norm.shape[1], 1)
y_train_cnn = y_balanced

print(f'CNN input shape: {X_train_cnn.shape}')
print('(n_stars, timesteps, channels)')
```

```python
# ── Build the 1D CNN architecture ──
def build_cnn(input_length):
    model = keras.Sequential([
        # Input layer
        layers.Input(shape=(input_length, 1)),
        
        # Block 1: Detect broad patterns
        layers.Conv1D(filters=32, kernel_size=11, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=4),
        layers.Dropout(0.2),
        
        # Block 2: Detect finer patterns
        layers.Conv1D(filters=64, kernel_size=7, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=4),
        layers.Dropout(0.2),
        
        # Block 3: Detect even finer patterns
        layers.Conv1D(filters=128, kernel_size=5, activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling1D(pool_size=4),
        layers.Dropout(0.3),
        
        # Flatten and classify
        layers.GlobalAveragePooling1D(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.4),
        layers.Dense(64, activation='relu'),
        layers.Dense(1, activation='sigmoid')  # output: probability of exoplanet
    ])
    
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model

cnn_model = build_cnn(X_train_cnn.shape[1])
cnn_model.summary()
```

```python
# ── Train the CNN ──
# Callbacks: early stopping and learning rate reduction
callbacks = [
    keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=8,          # stop if no improvement for 8 epochs
        restore_best_weights=True
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,          # halve the learning rate if stuck
        patience=4,
        min_lr=1e-6
    )
]

print('Training 1D CNN... (this may take a few minutes)')
history = cnn_model.fit(
    X_train_cnn, y_train_cnn,
    epochs=50,
    batch_size=32,
    validation_split=0.2,   # 20% of training data used for validation
    callbacks=callbacks,
    verbose=1
)
print('\nCNN training complete!')
```

```python
# ── Plot training history ──
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('CNN Training History', fontsize=14, fontweight='bold')

# Loss
ax1.plot(history.history['loss'], label='Training Loss', color='deepskyblue')
ax1.plot(history.history['val_loss'], label='Validation Loss', color='orange')
ax1.set_title('Loss over epochs')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()

# Accuracy
ax2.plot(history.history['accuracy'], label='Training Accuracy', color='lightgreen')
ax2.plot(history.history['val_accuracy'], label='Validation Accuracy', color='salmon')
ax2.set_title('Accuracy over epochs')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy')
ax2.legend()

plt.tight_layout()
plt.savefig('cnn_training_history.png', dpi=150, bbox_inches='tight')
plt.show()
print('Ideal: both curves should go down together (training and validation)')
print('If training goes down but validation goes up = overfitting')
```

---
## Step 7: Evaluate CNN on Test Data

```python
# ── Make predictions ──
cnn_pred_prob = cnn_model.predict(X_test_cnn).flatten()
cnn_pred      = (cnn_pred_prob > 0.5).astype(int)

print('='*50)
print('1D CNN RESULTS')
print('='*50)
print(classification_report(y_test, cnn_pred,
                              target_names=['No Planet', 'Exoplanet']))

cnn_roc = roc_auc_score(y_test, cnn_pred_prob)
cnn_f1  = f1_score(y_test, cnn_pred)
print(f'ROC-AUC Score: {cnn_roc:.4f}')
print(f'F1 Score:      {cnn_f1:.4f}')
```

```python
# ── Confusion matrix for CNN ──
fig, ax = plt.subplots(figsize=(6, 5))
cm_cnn = confusion_matrix(y_test, cnn_pred)
sns.heatmap(cm_cnn, annot=True, fmt='d', cmap='Greens',
            xticklabels=['No Planet', 'Exoplanet'],
            yticklabels=['No Planet', 'Exoplanet'])
ax.set_title('1D CNN — Confusion Matrix', fontsize=13, fontweight='bold')
ax.set_ylabel('Actual')
ax.set_xlabel('Predicted')
plt.tight_layout()
plt.savefig('cnn_confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── ROC Curves: RF vs CNN ──
fig, ax = plt.subplots(figsize=(8, 6))

# Random Forest ROC
fpr_rf, tpr_rf, _ = roc_curve(y_test, rf_pred_prob)
ax.plot(fpr_rf, tpr_rf, color='deepskyblue', linewidth=2,
        label=f'Random Forest (AUC = {rf_roc:.3f})')

# CNN ROC
fpr_cnn, tpr_cnn, _ = roc_curve(y_test, cnn_pred_prob)
ax.plot(fpr_cnn, tpr_cnn, color='lightgreen', linewidth=2,
        label=f'1D CNN (AUC = {cnn_roc:.3f})')

# Diagonal = random guessing
ax.plot([0, 1], [0, 1], 'r--', linewidth=1, label='Random guessing')

ax.set_xlabel('False Positive Rate', fontsize=12)
ax.set_ylabel('True Positive Rate', fontsize=12)
ax.set_title('ROC Curve: Random Forest vs 1D CNN', fontsize=13, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('roc_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('Higher curve = better model. AUC closer to 1.0 = better.')
```

## Step 8: Final Comparison & Interpretation

```python
# ── Final model comparison ──
print('='*55)
print('         FINAL MODEL COMPARISON')
print('='*55)
print(f'  Model               ROC-AUC    F1 Score')
print(f'  ─────────────────   ───────    ────────')
print(f'  Random Forest       {rf_roc:.4f}     {rf_f1:.4f}')
print(f'  1D CNN              {cnn_roc:.4f}     {cnn_f1:.4f}')
print('='*55)

better = 'CNN' if cnn_roc > rf_roc else 'Random Forest'
print(f'\n  Winner: {better}')
print()
print('  Why ROC-AUC matters more than accuracy here:')
print('  - Dataset is highly imbalanced (~37 planets out of 5000 stars)')
print('  - A model predicting NO PLANET always gets 99% accuracy but is useless')
print('  - ROC-AUC and F1 measure how well we actually DETECT the rare class')
```

```python
# ── Visualize: what does the CNN actually predict? ──
# Show some test stars with their predicted probabilities

fig, axes = plt.subplots(2, 3, figsize=(18, 8))
fig.suptitle('CNN Predictions on Test Stars', fontsize=14, fontweight='bold')

# Find actual exoplanet stars in test set
exo_idx  = np.where(y_test == 1)[0]
norm_idx = np.where(y_test == 0)[0]

# Top row: actual exoplanets
for i in range(3):
    idx  = exo_idx[i]
    prob = cnn_pred_prob[idx]
    color = 'lightgreen' if prob > 0.5 else 'tomato'
    verdict = '✅ DETECTED' if prob > 0.5 else '❌ MISSED'
    axes[0, i].plot(X_test_norm[idx], color='deepskyblue', linewidth=0.7)
    axes[0, i].set_title(f'ACTUAL: Exoplanet | CNN: {prob:.2%} → {verdict}',
                          color=color, fontsize=10)
    axes[0, i].set_ylabel('Normalized Flux')

# Bottom row: actual non-planets
for i in range(3):
    idx  = norm_idx[i]
    prob = cnn_pred_prob[idx]
    color = 'lightgreen' if prob <= 0.5 else 'tomato'
    verdict = '✅ CORRECT' if prob <= 0.5 else '❌ FALSE ALARM'
    axes[1, i].plot(X_test_norm[idx], color='tomato', linewidth=0.7)
    axes[1, i].set_title(f'ACTUAL: No Planet | CNN: {prob:.2%} → {verdict}',
                          color=color, fontsize=10)
    axes[1, i].set_ylabel('Normalized Flux')

for ax in axes.flat:
    ax.set_xlabel('Time step')

plt.tight_layout()
plt.savefig('cnn_predictions_visual.png', dpi=150, bbox_inches='tight')
plt.show()
```
